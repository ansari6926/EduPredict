# Data Visualization

Visualizing patterns in student data.

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../dataset/processed_data/cleaned_student_data.csv')
```

```
sns.countplot(data=df, x='Grade')
plt.title('Grade Distribution')
plt.show()
```

```
sns.boxplot(data=df, x='Department', y='Performance')
plt.title('Performance by Department')
plt.show()
```
