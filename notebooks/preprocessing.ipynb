# Data Preprocessing

This notebook handles missing values, outliers, and feature engineering.

```
import pandas as pd

df = pd.read_csv('../dataset/raw_data/student_data.csv')
df.isnull().sum()
```

```
# Feature Engineering: Total Score
df['TotalScore'] = df['Attendance'] * 0.3 + df['Internal1'] * 0.25 + df['Internal2'] * 0.25
df.head()
```

```
df.to_csv('../dataset/processed_data/cleaned_student_data.csv', index=False)
```
